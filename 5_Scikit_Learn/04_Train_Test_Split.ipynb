### Train-Test Split

Machine learning models should not be trained and tested on the same data.

The dataset is divided into:

- Training data
- Testing data

### Training Data

- Used to train the machine learning model.
- The model learns patterns from this data.
- Usually forms the larger portion of the dataset.

### Testing Data

- Used to evaluate the trained model.
- The model does not learn from this data.
- It helps measure how well the model performs on unseen data.

````python
from sklearn.datasets import load_iris
````

````python
from sklearn.model_selection import train_test_split
````

### train_test_split

- `train_test_split` is a Scikit-Learn function.
- It splits a dataset into training and testing sets.
- It is one of the most commonly used functions in supervised machine learning.

````python
iris = load_iris()
````

````python
X_train,X_test,y_train,y_test = train_test_split(
    iris.data,
    iris.target
)
````

````python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
````

### Understanding `X` and `y` Shapes

- `X` stores the input features (independent variables).
- `y` stores the target labels (dependent variable).
- Each sample has multiple feature values but only one target label.

Example:

```python
X.shape = (150, 4)
y.shape = (150,)
```

- `150` → Number of samples
- `4` → Number of features per sample
- `y` has shape `(150,)` because each sample has exactly one target label.

### Shapes After Train-Test Split

After splitting the dataset:

```python
X_train.shape
X_test.shape
y_train.shape
y_test.shape
```

The number of samples changes according to the split ratio, but:

- `X` always keeps the feature dimension.
- `y` always contains one target value for each sample.
- The number of samples in `X_train` and `y_train` is always the same.
- The number of samples in `X_test` and `y_test` is always the same.

### `random_state`

- `random_state` controls the randomness of the train-test split.
- Using the same `random_state` value produces the same train-test split every time.
- Different `random_state` values produce different train-test splits.
- If `random_state` is not specified, a different split may be generated each time the code is executed.

**Why use `random_state`?**

It makes experiments reproducible and allows different people to obtain the same train-test split.
