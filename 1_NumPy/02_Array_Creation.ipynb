# Array Creation

## Creating Arrays with `np.array()`

- `np.array()` creates a NumPy array from Python objects like lists or tuples.
- It is the most commonly used function to create arrays.

```python
import numpy as np
```

```python
A = np.array([10,20,30])
print(A)
print(type(A))
```

## Why NumPy Arrays?

Python lists are general-purpose containers.

NumPy arrays are designed for numerical computing.

### NumPy Advantages

- Supports element-wise mathematical operations.
- Faster than Python lists for numerical computations.
- Memory efficient.
- Foundation for libraries like Pandas, TensorFlow, PyTorch, and Scikit-learn.

```python
A = np.array([
    [10,20,30],
    [40,50,60]
])
print(A)
```

```python
A = np.array([
    [
        [1,2],
        [3,4]
    ],
    [
        [5,6],
        [7,8]
    ]  
])
print(A)
```

```python
A = np.array((10,20,30))
print(A)
```

```python
A = np.array([1,2,3], dtype = float)
print(A)
print(A.dtype)
```

3D array with one table:

```python
A = np.array([
    [
        [1,2],
        [3,4]
    ]
])
print(A)
```

```python
A = np.array((100,200,300))
print(A)
```

## np.zeros()

# Creating Arrays with `np.zeros()`

- `np.zeros()` creates a NumPy array filled with zeros.
- By default, the data type is `float64`.
- Useful for initialization in numerical computing and machine learning.

```python
A = np.zeros(5)
print(A)
print(A.dtype)
```

```python
A = np.zeros(10)
print(A)
```

```python
A = np.zeros((3,2))
print(A)
```

```python
A = np.zeros((2,2,2))
print(A)
```
