# Array Creation

## Creating Arrays with `np.array()`

- `np.array()` creates a NumPy array from Python objects like lists or tuples.
- It is the most commonly used function to create arrays.

````python
import numpy as np
````

````python
A = np.array([10,20,30])
print(A)
print(type(A))
````

## Why NumPy Arrays?

Python lists are general-purpose containers.

NumPy arrays are designed for numerical computing.

### NumPy Advantages

- Supports element-wise mathematical operations.
- Faster than Python lists for numerical computations.
- Memory efficient.
- Foundation for libraries like Pandas, TensorFlow, PyTorch, and Scikit-learn.

````python
A = np.array([
    [10,20,30],
    [40,50,60]
])
print(A)
````

````python
A = np.array([
    [
        [1,2],
        [3,4]
    ],
    [
        [5,6],
        [7,8]
    ]  
])
print(A)
````

````python
A = np.array((10,20,30))
print(A)
````

````python
A = np.array([1,2,3], dtype = float)
print(A)
print(A.dtype)
````

3D array with one table:

````python
A = np.array([
    [
        [1,2],
        [3,4]
    ]
])
print(A)
````

````python
A = np.array((100,200,300))
print(A)
````

## np.zeros()

# Creating Arrays with `np.zeros()`

- `np.zeros()` creates a NumPy array filled with zeros.
- By default, the data type is `float64`.
- Useful for initialization in numerical computing and machine learning.

````python
A = np.zeros(5)
print(A)
print(A.dtype)
````

````python
A = np.zeros(10)
print(A)
````

````python
A = np.zeros((3,2))
print(A)
````

````python
A = np.zeros((2,2,2))
print(A)
````

````python
A =np.zeros((4,5))
print(A)
````

````python
A = np.zeros((2,4,5))
print(A)
````

````python
A = np.zeros(6, dtype = int)
print(A)
````

## Creating Arrays with `np.ones()`

- `np.ones()` creates a NumPy array filled with ones.
- By default, the data type is `float64`.
- You can change the data type using the `dtype` parameter.

````python
A = np.ones(5)

print(A)
print(A.dtype)
````

````python
A = np.ones((2,3))

print(A)
````

````python
A = np.ones((2,2,2), dtype=int)

print(A)
print(A.dtype)
````

## Creating Arrays with `np.empty()`

- `np.empty()` creates an array without initializing its values.
- The array contains whatever data already exists in memory.
- It is faster than `zeros()` and `ones()` because NumPy does not fill the array.

````python
A = np.empty(5)
print(A)
````

````python
A = np.empty(5)
B = np.empty(5)
C = np.empty(5)

print(A)
print(B)
print(C)
````

## Creating Arrays with `np.full()`

- `np.full()` creates an array filled with a specified value.
- Unlike `zeros()` and `ones()`, you can choose any value.
- You can also specify the data type using `dtype`.
- np.full(shape, fill_value)

````python
A = np.full(3,4)
print(A)
````

````python
A = np.full((2,3), 5)
print(A)
````

````python
A = np.full((4,4),"Hello")
print(A)
````

## Creating Arrays with `np.arange()`

- `np.arange()` creates an array with evenly spaced values.
- Similar to Python's `range()`, but returns a NumPy array.
- Commonly used for indexing, loops, plotting, and machine learning.

### Syntax

```python
np.arange(stop)
np.arange(start, stop)
np.arange(start, stop, step)
```

### Rules

- Starts from `start` (default is `0`).
- Stops **before** `stop`.
- Moves by `step` (default is `1`).
- `step` can be positive, negative, or a float.

````python
A = np.arange(5)
print(A)
print(type(A))
````

````python
A = np.arange(2,10,2)
print(A)
````

````python
A = np.arange(0,10,2)
print(A)
````

````python
A = np.arange(5,30,5)
print(A)
````

````python
A = np.arange(10,4,-1)
print(A)
````

````python
A = np.arange(1, 2, 0.2)
print(A)
````

Start = 1....
Step = 0.2....
NumPy keeps adding 0.2

## Creating Arrays with `np.linspace()`

- `np.linspace()` creates evenly spaced values between a start and stop value.
- Unlike `np.arange()`, you specify the **number of values**, not the step size.
- By default, the stop value is included.

````python
A = np.linspace(0,10,6)
print(A)
````

````python
A = np.linspace(10,20,6)
print(A)
````

````python
A = np.linspace(0,100,5)
print(A)
````

````python
A = np.linspace(0,1,11)
print(A)
````

````python
A = np.linspace(100,200,5)
print(A)
````

 ### endpoint=True (default) and endpoint=False does not include the stop or last element.

````python
A = np.linspace(0,10,6, endpoint=False)
print(A)
````

## Creating Arrays with `np.eye()`

- `np.eye()` creates an identity matrix.
- An identity matrix has `1`s on the main diagonal and `0`s everywhere else.
- By default, the matrix is square (same number of rows and columns).

````python
A = np.eye(3)
print(A)
````

````python
A = np.eye(5)
print(A)
````

np.eye(rows, columns)

````python
A = np.eye(3,5)
print(A)
````

There are only 3 ones, because there are only 3 rows.

````python
A = np.eye(4,5)
print(A)
````

````python
A = np.eye(3,5)
print(A)
````

````python
A = np.eye(4, k=1)
print(A)
````

k = 1
Move the diagonal one column to the right.

k = -1 → move the diagonal one column to the left

````python
A = np.eye(4, k=-1)
print(A)
````

````python
A = np.eye(5)
print(A)
````
