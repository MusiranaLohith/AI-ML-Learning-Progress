# Introduction to NumPy

### What is NumPy?

- **NumPy (Numerical Python)** is a Python library for fast numerical computing.
- It provides a powerful data structure called **ndarray (N-dimensional Array)**.
- NumPy is faster and more memory-efficient than Python lists for numerical data.
- It is the foundation of libraries like **Pandas, Matplotlib, Scikit-learn, TensorFlow, and PyTorch**.

### Importing NumPy

```python
import numpy as np
```

- `np` is the standard alias used by the Python community.
- It makes the code shorter and easier to read.

````python
import numpy as np
print(np.__version__)
````

## Why NumPy?

Python lists are flexible but slower for numerical computations.

**Python List**
- Can store multiple data types
- Slower
- More memory

**NumPy Array**
- Stores one data type
- Faster
- Memory efficient

````python
numbers = [10,"Hello",3.14,True]
print(numbers)
````

````python
print(type(numbers))
````

````python
arr = np.array([10,20,30,])
print(arr)
print(type(arr))
````

ndarray = N-dimensional Array

## Array Dimensions

- 0D → Single value
- 1D → One-dimensional array (vector)
- 2D → Two-dimensional array (rows and columns)
- 3D → Collection of 2D arrays
- `ndim` returns the number of dimensions of an array.

````python
A = np.array([
    [10,20,30],
    [40,50,60]
])
print(A.shape)
````

````python
print(type(A.shape))
````

````python
A = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print(A.ndim)
print(A.shape)
````

````python
A = np.array([
    [
        [1, 2, 3, 4]
    ],
    [
        [5, 6, 7, 8]
    ]
])
print(A.ndim)
print(A.shape)
````

## NumPy Array Attributes

### `ndim`
- Returns the **number of dimensions** of an array.
- 1D → `(elements,)`
- 2D → `(rows, columns)`
- 3D → `(tables, rows, columns)`

### `shape`
- Returns the **size of each dimension** as a tuple.
- Count from the **outermost brackets inward**.

### `size`
- Returns the **total number of elements** in the array.

### `itemsize`
- Returns the **memory (in bytes)** occupied by **one element**.

### `nbytes`
- Returns the **total memory (in bytes)** occupied by the entire array.

> Formula:
>
> `nbytes = size × itemsize`

````python
A = np.array([10, 20, 30])

print(A.dtype)
````

````python
A = np.array([10, 20, 30])
print(A.dtype)
B = np.array([10.5, 20.8, 30.2])
print(B.dtype)
````

````python
A = np.array([10, 20, 30])

print("Array:", A)
print("ndim:", A.ndim)
print("shape:", A.shape)
print("size:", A.size)
print("itemsize:", A.itemsize)
print("nbytes:", A.nbytes)
print("dtype:", A.dtype)
````

````python
A = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:\n", A)
print("ndim:", A.ndim)
print("shape:", A.shape)
print("size:", A.size)
print("itemsize:", A.itemsize)
print("nbytes:", A.nbytes)
print("dtype:", A.dtype)
````

````python
A = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print("Array:\n", A)
print("ndim:", A.ndim)
print("shape:", A.shape)
print("size:", A.size)
print("itemsize:", A.itemsize)
print("nbytes:", A.nbytes)
print("dtype:", A.dtype)
````

## `len()` Function

- Returns the **size of the first (outermost) dimension** of a NumPy array.
- It is **not** the total number of elements.
- Equivalent to:

```python
len(A) == A.shape[0]
```

````python
import numpy as np

A = np.array([10, 20, 30])

print("Array:", A)
print("len(A):", len(A))
print("shape:", A.shape)
print("size:", A.size)
````

````python
A = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(A)

print("len(A):", len(A))
print("shape:", A.shape)
print("size:", A.size)
````
