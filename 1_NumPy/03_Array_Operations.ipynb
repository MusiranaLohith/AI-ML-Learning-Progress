# Arithmetic Operations on NumPy Arrays

- NumPy performs arithmetic operations element by element.
- Arrays must usually have the same shape.
- Operations are much faster than using Python loops.

## Basic Arithmetic Operations

NumPy performs arithmetic element by element.

Supported operations:

- `+` → Addition
- `-` → Subtraction
- `*` → Multiplication
- `/` → Division
- `**` → Power
- `%` → Modulus
- `//` → Floor Division

```python
import numpy as np

A = np.array([10, 20, 30])
B = np.array([2, 5, 10])

print("Addition      :", A + B)
print("Subtraction   :", A - B)
print("Multiplication:", A * B)
print("Division      :", A / B)
print("Power         :", A ** 2)
print("Modulus       :", A % B)
print("Floor Division:", A // B)
```

```python
A = np.array([10,20,30])

print(A + 5)
print(A - 5)
print(A % 5)
print(A * 5)
print(A / 5)
```

### NumPy broadcasts the single value 5 across the array internally.

### This is called broadcasting.

# Broadcasting

- Broadcasting allows NumPy to perform operations on arrays of different shapes.
- Instead of copying data, NumPy automatically stretches dimensions when possible.
- Broadcasting makes code shorter, faster, and avoids unnecessary loops.
